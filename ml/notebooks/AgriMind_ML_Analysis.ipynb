```python
# Section 2: Environment & Package Verification
import sys
import os
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
import tensorflow as tf

print(f"Python Version: {sys.version}")
print(f"TensorFlow Version: {tf.__version__}")
print(f"NumPy Version: {np.__version__}")
print(f"Pandas Version: {pd.__version__}")
print(f"Scikit-Learn Version: {sklearn.__version__}")
```

```python
# Section 3: Dataset Loading & Exploration
dataset_dir = "../dataset"
classes = [d for d in os.listdir(dataset_dir) if os.path.isdir(os.path.join(dataset_dir, d))]
print(f"Dataset Path: {os.path.abspath(dataset_dir)}")
print(f"Detected Classes: {classes}")

counts = {}
for c in classes:
    c_path = os.path.join(dataset_dir, c)
    imgs = [f for f in os.listdir(c_path) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
    counts[c] = len(imgs)
    print(f"Class '{c}': {len(imgs)} images")
```

```python
# Section 4: Dataset Visualization & Class Distribution
plt.figure(figsize=(8, 4))
plt.bar(counts.keys(), counts.values(), color=['#e74c3c', '#f39c12', '#2ecc71'])
plt.title('AgriMind AI - Class Distribution')
plt.xlabel('Crop Disease Class')
plt.ylabel('Number of Images')
plt.xticks(rotation=15)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Section 5: CNN Model Architecture Definition
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Rescaling(1./255, input_shape=(128, 128, 3)),
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(128, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.3),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.4),
    layers.Dense(len(classes) if len(classes) > 0 else 3, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()
```
